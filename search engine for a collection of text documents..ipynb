# Web Scraping

```python
# Setup headless Chrome
options = Options()
options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

# Function to get news links from BBC
def get_news_links(base_url, keyword, max_links=20):
    try:
        driver.get(base_url)
        time.sleep(3)
        links = set()
        elements = driver.find_elements(By.TAG_NAME, "a")
        for el in elements:
            href = el.get_attribute("href")
            if href and keyword in href and href.startswith("https://"):
                links.add(href)
            if len(links) >= max_links:
                break
        return list(links)
    except Exception as e:
        print(f"Error while getting links from {base_url}: {e}")
        return []

# Function to extract article text from a URL
def extract_article_text(url):
    try:
        driver.get(url)
        time.sleep(3)
        paragraphs = driver.find_elements(By.TAG_NAME, "p")
        content = ' '.join([p.text for p in paragraphs if p.text.strip() != ''])
        return content
    except Exception as e:
        print(f"Failed to extract text from {url}: {e}")
        return ""

# Open CSV file to write results
with open("bbc_scraped_news.csv", "w", newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(["URL", "Text"])

    # Scraping links from BBC only
    site = "https://www.bbc.com/news"
    keyword = "/news/"  # Specific to BBC news section
    print(f"\n Scraping links from: {site}")
    links = get_news_links(site, keyword, max_links=20)

    for link in links:
        print(f" Extracting article: {link}")
        text = extract_article_text(link)
        if text:
            writer.writerow([link, text])

driver.quit()
print("\n All articles scraped and saved to 'bbc_scraped_news.csv'")
```

# Pre-processing

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')

# Load the CSV file
df = pd.read_csv('bbc_scraped_news.csv')

lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    text = text.lower()

    text = re.sub(r'[^a-z\s]', '', text)

    tokens = word_tokenize(text)

    tokens = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words]

    return ' '.join(tokens)

df['Processed_Text'] = df['Text'].apply(preprocess_text)

df[['URL', 'Processed_Text']].to_csv('preprocessed_news.csv', index=False)

print("Preprocessing complete. Saved to 'preprocessed_news.csv'")
```

# Build corpus and Inverted Index

```python
def preprocess(text):

    tokens = word_tokenize(text.lower())

    tokens = [word for word in tokens if word.isalnum() and word not in stopwords.words("english")]

    return tokens


def build_corpus(scraped_data):
    corpus = []
    for doc in scraped_data:
        text = doc.get("Text", "")
        tokens = preprocess(text)
        corpus.append(tokens)
    return corpus


def build_inverted_index(corpus):
    inverted_index = defaultdict(list)

    for doc_id, tokens in enumerate(corpus):
        term_freq = defaultdict(int)
        for term in tokens:
            term_freq[term] += 1

        for term, freq in term_freq.items():
            inverted_index[term].append((doc_id, freq))

    return inverted_index

def read_csv(file_name):
    data = []
    with open(file_name, "r", encoding="utf-8") as file:
        reader = csv.reader(file)
        next(reader)
        for row in reader:
            data.append({"URL": row[0], "Text": row[1]})
    return data

csv_file = "preprocessed_news.csv"
scraped_data = read_csv(csv_file)

if scraped_data:
    corpus = build_corpus(scraped_data)
    print(f"Corpus built with {len(corpus)} documents.")

    inverted_index = build_inverted_index(corpus)
    print("\nInverted Index:")
    for term, doc_list in inverted_index.items():
        print(f"{term}: {doc_list}")
else:
    print("No data found to build the corpus and inverted index.")
```

# Calculate TF-IDF

```python
def load_documents_from_csv(file_path):
    df = pd.read_csv(file_path)
    corpus = df['Processed_Text'].tolist()
    return corpus, df

def preprocess(text):
    tokens = text.lower().split()
    return tokens

def build_inverted_index(corpus):
    inverted_index = defaultdict(list)
    for doc_id, doc in enumerate(corpus):
        tokens = preprocess(doc)
        term_freq = defaultdict(int)
        for term in tokens:
            term_freq[term] += 1
        for term, freq in term_freq.items():
            inverted_index[term].append((doc_id, freq))
    return inverted_index

def calculate_tfidf(inverted_index, corpus, query):
    query_tokens = preprocess(query)
    query_tf = defaultdict(int)
    for token in query_tokens:
        query_tf[token] += 1
    num_docs = len(corpus)
    idf = {}
    for term in query_tokens:
        doc_freq = len(inverted_index.get(term, []))
        idf[term] = math.log((num_docs + 1) / (doc_freq + 1)) + 1
    document_scores = defaultdict(float)
    for term, tf in query_tf.items():
        if term in inverted_index:
            for doc_id, term_freq in inverted_index[term]:
                tfidf_score = tf * idf[term] * term_freq
                document_scores[doc_id] += tfidf_score
    ranked_docs = sorted(document_scores.items(), key=lambda x: x[1], reverse=True)
    return ranked_docs

def process_query(inverted_index, corpus, df, query):
    ranked_docs = calculate_tfidf(inverted_index, corpus, query)
    print(f"\nRanked Documents for query: '{query}'\n")
    for doc_id, score in ranked_docs:
        url = df.loc[doc_id, 'URL']
        print(f"Document ID: {doc_id}, Score: {score}")
        print(f"URL: {url}")


corpus, df = load_documents_from_csv('preprocessed_news.csv')

inverted_index = build_inverted_index(corpus)

queries = [

    "climate change",

]

for query in queries:
    process_query(inverted_index, corpus, df, query)
```

# BM25

```python
import pandas as pd
from collections import defaultdict
import math

def load_documents_from_csv(file_path):
    df = pd.read_csv(file_path)
    corpus = df['Processed_Text'].tolist()
    return corpus, df

def preprocess(text):
    tokens = text.lower().split()
    return tokens

def build_inverted_index_and_doc_lengths(corpus):
    inverted_index = defaultdict(list)
    doc_lengths = []
    for doc_id, doc in enumerate(corpus):
        tokens = preprocess(doc)
        doc_lengths.append(len(tokens))
        term_freq = defaultdict(int)
        for term in tokens:
            term_freq[term] += 1
        for term, freq in term_freq.items():
            inverted_index[term].append((doc_id, freq))
    avg_doc_len = sum(doc_lengths) / len(doc_lengths)
    return inverted_index, doc_lengths, avg_doc_len

def calculate_bm25(inverted_index, corpus, query, doc_lengths, avg_doc_len, k1=1.5, b=0.75):
    query_tokens = preprocess(query)
    num_docs = len(corpus)
    document_scores = defaultdict(float)

    for term in query_tokens:
        if term in inverted_index:
            df = len(inverted_index[term])
            idf = math.log((num_docs - df + 0.5) / (df + 0.5) + 1)

            for doc_id, tf in inverted_index[term]:
                doc_len = doc_lengths[doc_id]
                denom = tf + k1 * (1 - b + b * doc_len / avg_doc_len)
                score = idf * tf * (k1 + 1) / denom
                document_scores[doc_id] += score

    ranked_docs = sorted(document_scores.items(), key=lambda x: x[1], reverse=True)
    return ranked_docs

def process_query(inverted_index, corpus, df, query, doc_lengths, avg_doc_len):
    ranked_docs = calculate_bm25(inverted_index, corpus, query, doc_lengths, avg_doc_len)
    print(f"\nRanked Documents for query: '{query}'\n")
    for doc_id, score in ranked_docs:
        url = df.loc[doc_id, 'URL']
        print(f"Document ID: {doc_id}, Score: {score:.4f}")
        print(f"URL: {url}\n")

corpus, df = load_documents_from_csv('preprocessed_news.csv')
inverted_index, doc_lengths, avg_doc_len = build_inverted_index_and_doc_lengths(corpus)

queries = ["climate change"]

for query in queries:
    process_query(inverted_index, corpus, df, query, doc_lengths, avg_doc_len)
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')

def preprocess(text):
    return text.lower().split()
def expand_query_with_embeddings(query, corpus, top_k_terms=3, top_k_similar=2, relevance_feedback=True):

    query_tokens = preprocess(query)
    query_vector = model.encode([query])[0]
    corpus_tokens = [preprocess(doc) for doc in corpus]
    corpus_vectors = model.encode(corpus)
    similarities = cosine_similarity([query_vector], corpus_vectors)[0]

    top_docs_indices = similarities.argsort()[-8:][::-1]
    top_docs = [(i, similarities[i]) for i in top_docs_indices]
    candidate_terms = []
    if relevance_feedback:
        for doc_id, _ in top_docs[:8]:
            candidate_terms.extend(corpus_tokens[doc_id])
    term_counts = Counter(candidate_terms)
    frequent_terms = [term for term, _ in term_counts.most_common(top_k_terms)]


    all_terms = list(set(candidate_terms))
    term_vectors = model.encode(all_terms)
    similarities = cosine_similarity([query_vector], term_vectors)[0]
    similar_term_indices = similarities.argsort()[-top_k_similar:][::-1]
    similar_terms = [all_terms[i] for i in similar_term_indices]


    expanded_query = list(set(query_tokens + frequent_terms + similar_terms))
    return ' '.join(expanded_query), top_docs
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')

def preprocess(text):

    return text.lower().split()


def expand_query_with_embeddings(query, corpus, top_k_terms=3, top_k_similar=2, relevance_feedback=True):

    query_tokens = preprocess(query)
    query_vector = model.encode([query])[0]
    corpus_tokens = [preprocess(doc) for doc in corpus]
    corpus_vectors = model.encode(corpus)
    similarities = cosine_similarity([query_vector], corpus_vectors)[0]


    top_docs_indices = similarities.argsort()[-5:][::-1]
    top_docs = [(i, similarities[i]) for i in top_docs_indices]
    candidate_terms = []
    if relevance_feedback:
        for doc_id, _ in top_docs[:5]:
            candidate_terms.extend(corpus_tokens[doc_id])
    term_counts = Counter(candidate_terms)
    frequent_terms = [term for term, _ in term_counts.most_common(top_k_terms)]


    all_terms = list(set(candidate_terms))
    term_vectors = model.encode(all_terms)

    similarities = cosine_similarity([query_vector], term_vectors)[0]
    similar_term_indices = similarities.argsort()[-top_k_similar:][::-1]
    similar_terms = [all_terms[i] for i in similar_term_indices]

    expanded_query = list(set(query_tokens + frequent_terms + similar_terms))
    return ' '.join(expanded_query), top_docs


def expand_multiple_queries(queries, corpus, top_k_terms=3, top_k_similar=2):
    expanded_queries = []
    all_top_docs = []

    for query in queries:
        expanded_query, top_docs = expand_query_with_embeddings(query, corpus, top_k_terms, top_k_similar)
        expanded_queries.append(expanded_query)
        all_top_docs.append(top_docs)

    return expanded_queries, all_top_docs
```

# Search engine

```python
def calculate_precision(retrieved_docs, relevant_docs):
    if not retrieved_docs:
        return 0.0
    relevant_retrieved = len(set(retrieved_docs) & set(relevant_docs))
    return relevant_retrieved / len(retrieved_docs)

def calculate_recall(retrieved_docs, relevant_docs):
    if not relevant_docs:
        return 0.0
    relevant_retrieved = len(set(retrieved_docs) & set(relevant_docs))
    return relevant_retrieved / len(relevant_docs)
```

```python
def search_engine(query, corpus, df, inverted_index, top_n=5):

    start_time = time.time()

    expanded_query, top_docs = expand_query_with_embeddings(query, corpus, top_k_terms=3, top_k_similar=2)

    ranked_docs = calculate_tfidf(inverted_index, corpus, expanded_query)

    results = []
    retrieved_doc_ids = []
    for doc_id, score in ranked_docs[:top_n]:
        if score >= 0:
            result = {
                'URL': df.iloc[doc_id]['URL'],
                'Text_Snippet': df.iloc[doc_id]['Processed_Text'][:200] + '...' if len(df.iloc[doc_id]['Processed_Text']) > 200 else df.iloc[doc_id]['Processed_Text'],
                'Score': score
            }
            results.append(result)
            retrieved_doc_ids.append(doc_id)

    relevant_docs = retrieved_doc_ids
    precision = calculate_precision(retrieved_doc_ids, relevant_docs)
    recall = calculate_recall(retrieved_doc_ids, relevant_docs)

    end_time = time.time()
    elapsed_time = end_time - start_time

    return results, expanded_query, elapsed_time, precision, recall
```

```python
print("BBC News Search Engine with Query Expansion and TF-IDF")

corpus, df = load_documents_from_csv('preprocessed_news.csv')
if not corpus or df is None:
    exit()

inverted_index = build_inverted_index(corpus)

query = input("Enter your search query: ")

results, expanded_query, search_time, precision, recall = search_engine(query, corpus, df, inverted_index, top_n=5)

print(f"\nExpanded Query: {expanded_query}")
print(f"\nSearch completed in {search_time:.4f} seconds")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")

if not results:
    print("No relevant articles found.")
else:
    print("\nSearch Results:")
    for i, result in enumerate(results, 1):
        print(f"\nResult {i}:")
        print(f"URL: {result['URL']}")
        print(f"Snippet: {result['Text_Snippet']}")
        print(f"Score: {result['Score']:.4f}")
```
